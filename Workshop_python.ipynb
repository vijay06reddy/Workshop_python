Loop Control Statement

```python
i = 1 #initialization
while(i<=5):  #while(condition)
  print("Hello, SRU")   #code that you want to repeat
  i=i+1   #increment variable value
```

```python
num=input("Enter the number")
num=int(num)
i = 0   #initialization
while (i<=9):
  print(f"{num} * {i+1} = {num*(i+1)}")
  i = i+1


```

Break Statement

```python
num=input("Enter the number")
num=int(num)
i = 1  #initialization
while (i<=10):
  print(f"{num} * {i} = {num*(i)}")
  if(i==5):
    break     #break statement
  i = i+1
```
