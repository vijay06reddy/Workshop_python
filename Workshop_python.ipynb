```python
num=input("Enter the number")
num=int(num)
i = 0   #initialization
while (i<=9):
  print(f"{num} * {i+1} = {num*(i+1)}")
  i = i+1


```

Break Statement

```python
num=input("Enter the number")
num=int(num)
i = 1  #initialization
while (i<=10):
  print(f"{num} * {i} = {num*(i)}")
  if(i==5):
    break     #break statement
  i = i+1
```

```python
List_of_books=["Python", "Java", "C", "C++"]
book=input("Enter book you want in library")
book_no=0 #start from index 0
while(book_no < len(List_of_books)):    #Loop condition corrected to avoid out of bounds access
  if(book==List_of_books[book_no]):
    print("Yes, book available")
    break
  book_no=book_no+1
```

While-Else Statement

```python
List_of_books=["Python", "Java", "C", "C++"]
book=input("Enter book you want in library")
book_no=0 #start from index 0
while(book_no < len(List_of_books)):    #Loop condition corrected to avoid out of bounds access
  if(book==List_of_books[book_no]):
    print("Yes, book available")
    break
  book_no=book_no+1
else:
    print("book not available")
```

Nested For loops

```python
for i in range(5):
  for j in range(i+1):
    print("*", end="")
  print()
```

```python
for i in range(3):
  for j in range(3):
    if(i!=j):
      print(i,j)
  print("=========")
```

Break statement

```python
for i in range(3):
  for j in range(3):
    if(i!=j):
      print(i,j)
      break
  print("======")
```

```python
for i in range(3):
  for j in range(3):
    if(i!=j):
      print(i,j)
  if(i==2):
    break
  print("======")
```

For-Else statement

```python
List_of_books=["Python", "Java", "C", "C++"]
book=input("Enter book you want in library")
no_of_books=len(List_of_books)
for book_no in range(no_of_books):
  if(book==List_of_books[book_no]):
    print("Yes, book available")
    break
else:
  print("No, book unavailable")
```
